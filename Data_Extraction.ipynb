```python
import pdfplumber
import re

# Function to extract values from multiple PDFs based on keywords
def extract_values_from_pdfs(pdf_files, keywords):
    extracted_values = []  # List to store all extracted values
    
    for pdf_file in pdf_files:
        with pdfplumber.open(pdf_file) as pdf:
            first_page = pdf.pages[0]
            text = first_page.extract_text()  # Extract text from the first page
            
            for keyword in keywords:
                if keyword == "total":
                    pattern = r"Total\s+USD\$\s*([0-9,.]+)"
                elif keyword == "gross amount":
                    pattern = r"Gross Amount incl\. VAT\s+([0-9,.]+)\s?€"
                else:
                    return("Keyword not Found!") 
                
                # Search for the pattern in the extracted text
                match = re.search(pattern, text)
                if match:
                    extracted_values.append(match.group(0))  # Add the matched value to the list

    return extracted_values

# Variables
pdf_files = ["sample_invoice_1.pdf", "sample_invoice_2.pdf"]  # List of PDF files to process
keywords = ["gross amount", "total"]  # List of keywords to search for

# Extract values for the given keywords from the provided PDFs
values = extract_values_from_pdfs(pdf_files, keywords)

# Display the extracted values
for value in values:
    print(value)
```
